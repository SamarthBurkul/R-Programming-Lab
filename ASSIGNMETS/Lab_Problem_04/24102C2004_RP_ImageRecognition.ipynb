```r
# 2. Resize all images to 28x28x3
for (i in 1:12) {
  my_pic[[i]] <- resize(my_pic[[i]], 28, 28)
}
```

```r
# 3. Reshape each 28x28x3 image into a 1D vector (length = 28 * 28 * 3 = 2352)
for (i in 1:12) {
  my_pic[[i]] <- array_reshape(my_pic[[i]], c(28, 28, 3))
}
```

```r
# 4. Form train_x (Images 1-5 for planes, 7-11 for cars)
train_x <- NULL
for (i in 1:5) {
  train_x <- rbind(train_x, my_pic[[i]])
}
for (i in 7:11) {
  train_x <- rbind(train_x, my_pic[[i]])
}
```

```r
# 5. Form test_x (Image 6 for plane, Image 12 for car)
test_x <- rbind(my_pic[[6]], my_pic[[12]])
```

```r
# 6. Create Binary Labels (0 = airplane, 1 = car)
train_y <- c(rep(0, 5), rep(1, 5))
test_y  <- c(0, 1)
```

```r
# One-hot encoding in pure R (no Python/TensorFlow required)
one_hot <- function(y, num_classes = 2) {
  t(sapply(y, function(x) {
    vec <- numeric(num_classes)
    vec[x + 1] <- 1
    vec
  }))
}

train_labels <- one_hot(train_y, num_classes = 2)
test_labels  <- one_hot(test_y, num_classes = 2)

print(train_labels)
print(test_labels)
```

```r
# Define the Sequential Architecture with the layer list
model <- keras_model_sequential(layers = list(
  layer_dense(units = 256, activation = "relu", input_shape = c(2352)),
  layer_dense(units = 128, activation = "relu"),
  layer_dense(units = 2, activation = "softmax")
))

# Inspect model summary
summary(model)
```

```r
model$compile(
  loss = "binary_crossentropy",
  optimizer = "rmsprop",
  metrics = list("accuracy")
)
```

```r
history <- model$fit(
  x = train_x,
  y = train_labels,
  epochs = as.integer(30),
  batch_size = as.integer(32),
  validation_split = 0.2
)

# Plot training history
plot(history$history$loss, type = "l", col = "blue", ylim = c(0, 1),
     xlab = "Epoch", ylab = "Value", main = "Model Loss & Accuracy")
lines(history$history$accuracy, col = "red")
legend("topright", legend = c("Loss", "Accuracy"), col = c("blue", "red"), lty = 1)
```

```r
# 1. Training Evaluation
train_loss_acc <- model$evaluate(train_x, train_labels)
cat("\nTrain Loss:", train_loss_acc[[1]], "Train Accuracy:", train_loss_acc[[2]], "\n")

train_prob <- model$predict(train_x)
train_pred <- apply(train_prob, 1, which.max) - 1

cat("\n--- Training Set Confusion Matrix ---\n")
table(Predicted = train_pred, Actual = train_y)

# 2. Test Evaluation
test_loss_acc <- model$evaluate(test_x, test_labels)
cat("\nTest Loss:", test_loss_acc[[1]], "Test Accuracy:", test_loss_acc[[2]], "\n")

test_prob <- model$predict(test_x)
test_pred <- apply(test_prob, 1, which.max) - 1

cat("\n--- Test Set Confusion Matrix ---\n")
table(Predicted = test_pred, Actual = test_y)
```

```r
# 1. Training Set Probabilities Table
train_summary <- data.frame(
  Image = pics[c(1:5, 7:11)],
  Prob_Airplane = round(train_prob[, 1], 4),
  Prob_Car      = round(train_prob[, 2], 4),
  Predicted     = ifelse(train_pred == 0, "Airplane", "Car"),
  Actual        = ifelse(train_y == 0, "Airplane", "Car")
)
print(train_summary)

# 2. Test Set Probabilities Table
test_summary <- data.frame(
  Image = pics[c(6, 12)],
  Prob_Airplane = round(test_prob[, 1], 4),
  Prob_Car      = round(test_prob[, 2], 4),
  Predicted     = ifelse(test_pred == 0, "Airplane", "Car"),
  Actual        = ifelse(test_y == 0, "Airplane", "Car")
)
print(test_summary)
```
