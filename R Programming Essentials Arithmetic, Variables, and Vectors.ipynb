Vector Addition

```r
# Equal length vectors
c(2, 3, 5, 7) + c(-2, -3, -5, 8)
```

```r
# Unequal length vectors (Recycling Rule)
c(2, 3, 5, 7) + c(8, 9)

```

```r
# Mismatched lengths (Generates Warning)
c(2, 3, 5, 7) + c(8, 9, 10)
```

```r
# Unequal length vectors
c(12, 13, 15, 17) - c(8, 9)
```

```r
# Mismatched lengths (Generates Warning)
c(12, 13, 15, 17) - c(8, 9, 10)
```

Basic Operations with Scalars

```r
# Addition
2 + 3

# Multiplication
2 * 3

# Subtraction
2 - 3

# Division
3 / 2

# Complex arithmetic expression
2 * 3 - 4 + 5 / 6
```

BODMAS Expressions

```r
# BODMAS order of operations
(2 + 3) * 5 + 5 - 10

# Nested parentheses calculation
(((2 + 3) * 5 + 5) - 10) / 2
```

Vector versus Scalar Operations

```r
# Addition with scalar
c(2, 3, 5, 7) + 10

# Subtraction with scalar
c(12, 13, 15, 17) - 10

# Multiplication with scalar
c(2, 3, 5, 7) * 10

# Division with scalar
c(12, 13, 15, 17) / 10
```

Power Operators with Scalars

```r
# Power using ^ and ** operators
2^3
2**3

# Fractional and negative powers
2^0.5
2**0.5
2^-0.5
```

Power Operations with Data Vectors

```r
# Applying power to a vector with a scalar
c(2, 3, 5, 7)^2

# Vector power with recycling rule
c(2, 3, 5, 7)^c(2, 3)

# Power operation with longer vector
c(1, 2, 3, 4, 5, 6)^c(2, 3, 4)

# Unequal length power operation (Generates Warning)
c(2, 3, 5, 7)^c(2, 3, 4)
```

Integer Division (%/%)

```r
# Integer division with scalars
2 %/% 2
5 %/% 2
7 %/% 3

# Integer division with vector and scalar
c(2, 3, 5, 7) %/% 2

# Integer division with vector and vector
c(2, 3, 5, 7) %/% c(2, 3)

# Unequal length integer division (Generates Warning)
c(2, 3, 5) %/% c(2, 3)
```

Modulo Division (%%)

```r
# Modulo division with scalars (Remainder)
2 %% 2
3 %% 2
7 %% 3
7 %% 4

# Modulo division with vector and scalar
c(2, 3, 5, 7) %% 2

# Modulo division with vector and vector
c(2, 3, 5, 7) %% c(2, 3)

# Unequal length modulo division (Generates Warning)
c(2, 3, 5) %% c(2, 3)
```

Maximum, Minimum, and Mean Functions

```r
# Maximum value
max(1.2, 3.4, -7.8)
max(c(1.2, 3.4, -7.8))

# Minimum value
min(1.2, 3.4, -7.8)
min(c(1.2, 3.4, -7.8))

# Mean value
mean(2, 3, 4)
mean(c(2, 3, 4)) # Correct vector usage
```
