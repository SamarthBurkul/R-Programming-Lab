## Lecture 40 — Substitution: `sub` and `gsub`

`sub(old, new, string)` replaces the **first** match; `gsub(old, new, string)` replaces **all** of them.

### 1. Simple replacement

```python
%%R
y = "Number of participants: 25"
sub("25", "30", y)
```

### 2. `sub` replaces only the first instance

```python
%%R
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."
y
sub("Mr. Singh", "Professor Jha", y)
```

### 3. `gsub` replaces every instance (global replace)

```python
%%R
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."
gsub("Mr. Singh", "Professor Jha", y)

# Recall
sub("Mr. Singh", "Professor Jha", y)
```

## Lecture 40 — Searching: `grep`

`grep` = **G**lobally search a **R**egular **E**xpression and **P**rint it.
`sub`/`gsub` replace; `grep`/`grepl` only search.

### 4. `value = TRUE` returns the matching elements themselves

```python
%%R
str = c("R Course", "exercises", "include examples of R language")
grep("ex", str, value=T)
```

### 5. `value = FALSE` (the default) returns the indices

```python
%%R
str = c("R Course", "exercises", "include examples of R language")
grep("ex", str, value=F)
```

### 6. `ignore.case` — case-sensitive vs case-insensitive

```python
%%R
str = c("R Course", "exercises", "include examples of r language", "in R software.")
grep("R", str, ignore.case=F, value=T)
grep("R", str, ignore.case=T, value=T)
```

### 7. Same comparison, returning indices

```python
%%R
str = c("R Course", "exercises", "include examples of r language", "in R software.")
grep("R", str, ignore.case=T, value=F)
grep("R", str, ignore.case=F, value=F)
```

### 8. Which string contains the pattern?

`"our"` sits inside `"course"`, so it is found in the 1st element only.

```python
%%R
x = "R course 24.07.2021"
y = "Number of participants: 25"
c(x, y)          # combine the two strings
grep("our", c(x,y))
```

### 9. And the other way round — `"Num"` is only in `y`

```python
%%R
x = "R course 24.07.2022"
y = "Number of participants: 50"
c(x, y)
grep("Num", c(x,y))
```

## Lecture 43 — Some More Operations

### 19. `summary` on a categorical variable gives a frequency table

```python
%%R
summary(painters$School)
```

### 20. `attach()` — reference variables directly by name

After `attach(painters)` the names `Composition`, `Drawing`, `Colour`, `Expression`,
`School` can be used without the `painters$` prefix.

```python
%%R
attach(painters)
summary(School)        # character/factor variable
summary(Composition)   # numeric variable
```

### 21. `detach()` restores the default — the bare name is gone again

```python
%%R
detach(painters)
summary(School)
```

### 22. Subsets with `subset()`

```python
%%R
subset(painters, School=='F')   # == is the logical equal sign
```

### 23. The equivalent indexing form

```python
%%R
painters[ painters[["School"]] == "F", ]
```

### 24. Subsetting on a numeric condition

```python
%%R
subset(painters, Composition <= 6)
```

### 25. Dropping uninteresting columns with `select`

```python
%%R
subset(painters, School=="F", select=c(-3,-5))   # drop Colour and School
```

### 26. `split` partitions the data set by a factor variable

```python
%%R
splitted = split(painters, painters$School)
splitted
```

### 27. Each piece is itself a data frame

```python
%%R
is.data.frame(splitted$A)
names(splitted)
```

## Lecture 44 — Combining and Merging

Three main techniques:

- `cbind()` — combine columns side-by-side
- `merge()` — join two data frames on a common column
- `rbind()` — stack two data frames on top of each other

### 28. Create the two example data frames

```python
%%R
df1 = data.frame(state=c("UP", "MP", "AP", "JK"),
                 popnsize=c(1000,2000,3000,4000))

df2 = data.frame(state=c("UP", "MP", "AP", "JK"),
                 samplesize=c(100,200,300,400),
                 surveycompleted=c("Yes", "No", "Yes", "No"))
df1
df2
```

### 29. `cbind` — note the duplicated `state` column

```python
%%R
df1 = data.frame(state=c("UP", "MP", "AP", "JK"),
                 popnsize=c(1000,2000,3000,4000))
df2 = data.frame(state=c("UP", "MP", "AP", "JK"),
                 samplesize=c(100,200,300,400),
                 surveycompleted=c("Yes", "No", "Yes", "No"))
cbind(df1, df2)
```

### 30. `merge` — joins on the common column and sorts by it

```python
%%R
df1 = data.frame(state=c("UP", "MP", "AP", "JK"),
                 popnsize=c(1000,2000,3000,4000))
df2 = data.frame(state=c("UP", "MP", "AP", "JK"),
                 samplesize=c(100,200,300,400),
                 surveycompleted=c("Yes", "No", "Yes", "No"))
merge(df1, df2, by="state")
```

### 31. `rbind` — stacking vertically (same columns required)

```python
%%R
df11 = data.frame(state=c("UP", "MP", "AP", "JK"),
                  popnsize=c(1000,2000,3000,4000))
df22 = data.frame(state=c("Bihar", "Delhi", "Punjab"),
                  popnsize=c(100,200,300))
df11
df22
rbind(df11, df22)
```

## Quick reference

| Function | Returns | Notes |
|---|---|---|
| `sub(old, new, x)` | character | replaces the **first** match |
| `gsub(old, new, x)` | character | replaces **all** matches |
| `grep(pat, x)` | integer indices | `value=TRUE` returns the elements instead |
| `grep(pat, x, ignore.case=TRUE)` | — | case-insensitive matching |
| `grepl(pat, x)` | logical vector | one `TRUE`/`FALSE` per element |
| `data.frame()` | data frame | mixes numeric, character and factor columns |
| `attach()` / `detach()` | — | use column names without the `df$` prefix |
| `subset(df, cond, select=)` | data frame | `select=c(-3,-5)` drops columns |
| `split(df, f)` | list of data frames | partition by a factor |
| `cbind` / `merge` / `rbind` | data frame | side-by-side / join on key / stack |
