```r
#power operator
2^3
2**3
2^0.5
2**0.5
2^-0.5

#power operator with scalar
c(2,3,5,7)^2

#power operator with vector
c(2,3,5,7)^c(2,3)

c(1,2,3,4,5,6)^c(2,3,4)

c(2,3,5,7)^c(2,3,4)
```

```r
#integer division with scalar
#division in which the fractional part(remainder) is discarded

2 %/% 2
7 %/% 3


#integer division with scalar
c(2,3,4,5,6) %/% 2

#integer division with data vector
c(2,3,5,7) %/% c(2,3)
```

```r
#modulo div with scalars
2 %% 2
7 %% 3

c(2,3,5,7) %% 2

c(2,3,5,7) %% c(2,3)

c(2,3,5) %% c(2,3)
```

```r
#functions
#max
max(1.2, 3.4, -7.8)
max(c(1.2,3.4,-7.8))

#min
min(1.2, 3.4, -7.8)
min(c(1.2,3.4,-7.8))

#mean
mean(2,3,4)
mean(c(2,3,4))

#abs
abs(-4)
abs(c(-1,-2,-3,-4,5))

#sqrt
sqrt(4)
sqrt(c(4,9,16,26,36))

#sum
sum(5,10,15)
sum(c(5,10,15))

#round
round(1.23)
round(1.83)

#log
log(10)
log(exp(1))
log(c(10,100,1000))

#log10
log10(10)
log10(100)
```
