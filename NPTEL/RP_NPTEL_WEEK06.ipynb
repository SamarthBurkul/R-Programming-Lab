## 1. `for` loop – basic example

```r
for ( i in 1:5 ) { print( i^2 ) }
```

## 2. `for` loop over a custom vector

```r
for ( i in c(2,4,6,7) ) { print( i^2 ) }
```

## 3. `for` loop with a counter using `if` inside a function

```r
x = c(2,4,6,8,10,12)

excount = function(x){
  count = 0
  for (xval in x) {
    if(xval/2 > 3)
      count = count+1
  }
  print(count)
}
excount(x)
```

## 4. Nested `for` loops

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

## 5. The `break` command

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}
```

## 6. The `next` command

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }
  print(x)
}
```

## 7. The `while` loop – basic example

```r
i = 1
while (i < 10) {
  print(i^2)
  i = i+2
}
```

## 8. The `while` loop – summing numbers via user input

```r
sumfunction = function(){
  sum = 0
  number = as.integer(readline(prompt="Please select any number less than 25: "))
  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the While Loop: ", sum))
}
sumfunction()
```

## 9. The `repeat` loop with `break`

```r
i = 1
repeat{
  print( i^2 )
  i = i+2
  if ( i > 10 )
    break
}
```

## 10. The `repeat` loop with `next` and `break`

```r
i = 1
repeat{
  i = i+1
  if (i < 10) next
  print(i^2)
  if (i >= 13) break
}
```

## 11. Functions – single variable

```r
abc = function(x){
  x^2
}
abc(3)
abc(6)
abc(9)
```

## 12. Functions – two variables

```r
abc = function(x,y){
  x^2+y^2
}
abc(3,4)
abc(10, 10)
abc(-2, -3)
```

## 13. Functions – using built-in math functions

```r
abc = function(x){
  sin(x)^2+cos(x)^2 + x
}
abc(9)
abc(99)
abc(-15)
```

## 14. Functions – calling a function without arguments (contains a `for` loop)

```r
abc = function() {
  for(i in 1:3) {
    print(i^3)
  }
}
abc()
```
