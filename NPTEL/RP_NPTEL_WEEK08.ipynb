## Create two lists (numeric and character)

```r
list1 = list(1,2,3)
list2 = list("water", "juice", "lemonade")
list1
list2
```

## Merge two lists using c()

```r
list12 = c(list1, list2)
list12
```

## Convert a list to a vector using unlist()

```r
unlist(list1)
unlist(list2)
mode(list1)
mode(unlist(list1))
```

## Append an element to the end of a list using append()

```r
append(list1, 100)
append(list2, "coffee")
```

## Append an element after a specific position using append(x, values, after)

```r
append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

## Remove an element from a list using listname[-position]

```r
list1[-2]
list2[-1]
```

```r
list2 = list("water", "juice", "lemonade", "tea", "coffee", "milk")
list2
list2[2:4]
list2[c(1,3,5)]
```

## Create a vector and view its values

```r
x = 1:10
x
```

## Subset a vector using a logical condition (x > 5)

```r
x[ (x > 5) ]
```

## Subset a vector for even values using the modulus operator %%

```r
x[ (x%%2==0) ]  #%% indicates x mod y, values for which x mod 2 is 0
```

## Subset a vector for odd values using the modulus operator %%

```r
x[ (x%%2==1) ]  #values for which x mod 2 is 1
```

```r
y = x[ !is.na(x) ]  #! means negation
y  # 5 is missing
```

```r
mean(x)
mean(y)
```

## Encode a numeric vector as a factor with custom levels and labels (die-roll example)

```r
y = c(1, 4, 3, 5, 4, 2, 4)
y
possible.dieface = c(1, 2, 3, 4, 5, 6)
possible.dieface
labels.dieface = c("one", "two", "three", "four", "five", "six")
labels.dieface
facy = factor(y, levels = possible.dieface, labels = labels.dieface)
facy
```

## Convert a numeric vector into a factor using as.factor() (levels ordered numerically)

```r
x = c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)
x
y = as.factor(x)
y
```

## Create a factor from a character vector (levels ordered alphabetically)

```r
x = factor( c("lemonade", "lemonade", "juice", "lemonade", "water") )
x
```

## Create factors from a character vector of brand labels

```r
brands = c("A","A","B","B","B","B","C")
brands
brands_fac = factor(brands)
brands_fac
```

## Use unclass() to view the underlying integer codes and levels attribute

```r
unclass(brands_fac)
```

## Use the unclassed integer codes to index into another vector (map factor codes to colours)

```r
colours = c("blue", "green", "red")
colours
unclass(brands_fac)
colours[unclass(brands_fac)]
```

## Unclass a character-based factor to see its integer codes and levels

```r
x = factor( c("lemonade", "lemonade", "juice", "lemonade", "water") )
unclass(x)
```

## Create a factor with a custom (non-alphabetical) level order using the levels argument

```r
x = factor( c("lemonade", "lemonade", "juice", "lemonade", "water"), levels=c("water", "juice", "lemonade") )
x
```

## Inspect the custom-ordered factor with unclass() and levels()

```r
unclass(x)
levels(x)
```

## Create an ordered factor using ordered() to represent ranked categories

```r
income = ordered(c("high", "high", "low", "medium", "medium"), levels=c("low", "medium", "high"))
income
unclass(income)
```

## Format a number with specified digits and nsmall (trailing decimal places)

## Format a character vector with fixed width and centre justification

```r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "centre")
format(c("A", "BB", "CCC", "DDDD"), width = 14, justify = "centre")
```

## Compare justify options (centre, left, right, none) with format()

```r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "centre")
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "left")
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "right")
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "none")
```
