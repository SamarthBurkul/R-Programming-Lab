## Sequences

`:` generates a continuous sequence with a constant unit increment, ascending or descending.

```r
1:10
10:1
5:15
15:5
```

```r
-1:-10
-10:-1
-5:-15
-15:-5
```

The `:` operator also works with decimal (non-integer) start and end values.

```r
1.23:10
1.23:10.54
10.54:2.23
```

```r
-1.23:-10
-5.23:6
```

`seq(n)` is a shorthand equivalent to `seq(1:n)`, generating 1 through n.

```r
seq(10)
seq(1:10)
```

`seq(from, to, by)` builds a sequence using a predefined variable to control the start, end, and step size.

```r
x = 2
seq(1, x, x/10)

x = 50
seq(0, x, x/10)
```

The output of `seq()` can be stored in a variable and then reused in further computations.

```r
x = seq(1, 50, 1/2)
x
y = 2*x
y
```

`seq(along = x)` creates an index vector matching the length of `x`, which can then be used to access elements of `x`.

```r
x = c(9,8,7,6)
ind = seq(along = x)
ind
x[ ind[2] ]
```

`Sys.time()` and `Sys.Date()` return the current date/time and current date from the system clock.

```r
Sys.time()
Sys.Date()
```

`seq()` on `Date` objects with `by = "years"` generates a sequence of first-of-year dates between a start and end date.

```r
seq(as.Date("2010-01-01"), as.Date("2017-01-01"), by = "years")
```

Using `by = "days"` with `length` generates a sequence of consecutive daily dates of a given length.

```r
seq(as.Date("2017-01-01"), by = "days", length = 6)
```

Using `by = "months"` generates a sequence of dates incremented one month at a time.

```r
seq(as.Date("2017-01-01"), by = "months", length = 6)
```

Using `by = "years"` with `length` generates a sequence of dates incremented one year at a time.

```r
seq(as.Date("2017-01-01"), by = "years", length = 6)
```

A date sequence can also be generated backward between an explicit start and end date using a negative step (e.g. `"-1 month"`).

```r
startdate = as.Date("2016-1-1")
enddate = as.Date("2017-1-1")
out = seq(enddate, startdate, by = "-1 month")
out
```

`letters` is a built-in constant giving the sequence of lowercase English alphabets.

```r
letters
```

`letters[from_index:to_index]` extracts a sub-sequence of lowercase alphabets by position.

```r
letters[1:3]
letters[3:1]
letters[21:23]
letters[2]
```

`LETTERS` is a built-in constant giving the sequence of uppercase English alphabets.

```r
LETTERS
```

`LETTERS[from_index:to_index]` extracts a sub-sequence of uppercase alphabets by position.

```r
LETTERS[1:3]
LETTERS[3:1]
LETTERS[21:23]
LETTERS[2]
```

## Repeats

`rep(x, times = n)` repeats an entire object `n` times, one full copy after another.

```r
rep(3.5, times = 10)
rep(1:4, 2)
```

`rep(x, each = n)` repeats every single element `n` times before moving to the next element.

```r
x = 1:4
x
rep(x, times = 3)
rep(x, each = 3)
```

Combining `each` and `times` shows that the order in which they are applied changes the resulting pattern (though here both orders give the same result since they are independent arguments).

```r
rep(1:4, each = 2)
rep(1:4, each = 2, times = 3)
rep(1:4, times = 3, each = 2)
```

Passing a vector to `times` repeats each corresponding element a different number of times.

```r
rep(1:4, 2:5)
```

The repeat-count vector itself can be generated dynamically, e.g. with `seq()`, and then used inside `rep()`.

```r
ans = seq(from = 2, to = 8, by = 2)
ans
rep(1:4, ans)
```

`rep()` also works on non-vector objects such as matrices, repeating their elements in column-major order.

```r
x = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = TRUE)
x
rep(x, 3)
```

`rep()` works equally well on character vectors, repeating the whole sequence of strings.

```r
rep(c("a", "b", "c"), 2)
rep(c("apple", "banana", "cake"), 2)
```

`rep(x, length.out = n)` (or `length = n`) repeats values cyclically until the output reaches the specified length.

```r
rep(2, length.out = 5)
rep(2, length = 5)
rep(c(2,3), length = 5)
rep(c(2,3,4), length = 5)
```

The `length` argument works the same way for character vectors, cycling through values until the target length is reached.

```r
rep("apple", length = 5)
rep(c("a", "b", "c"), length = 2)
rep(c("a", "b", "c"), length = 5)
```

## Sorting

`sort()` arranges the values of a vector in ascending order by default, or descending order with `decreasing = TRUE`.

```r
y = c(8,5,7,6)
y
sort(y)
sort(y, decreasing = TRUE)
```

## Ordering

`order()` returns the positions (indices) that would arrange the vector in sorted order, rather than the sorted values themselves.

```r
y = c(9,8,5,7,6)
y
order(y)
order(y, decreasing = TRUE)
```

## Mode

`mode()` reports how an R object is stored internally, e.g. numeric, character, list, or function.

```r
mode(2.432)
mode(c(3,4,5,6,7,8))
mode("India")
mode(c("India", "CANADA"))
```

Factors, lists, and data frames report their mode according to their internal storage type, and functions report mode `"function"`.

```r
mode(factor(c("UP", "MP")))
mode(list("India", "USA"))
mode(data.frame(x = 1:2, y = c("India", "USA")))
mode(print)
```

## Lists

Ordinary vectors and matrices can only hold one data type, so adding two numeric matrices works via elementwise arithmetic.

```r
x1 = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = TRUE)
x2 = matrix(nrow = 2, ncol = 2, data = 5:8, byrow = TRUE)
x1
x2
x1 + x2
```

Assigning a character value into a numeric matrix coerces the whole matrix to character mode, so arithmetic on it then fails.

```r
x1[2,1] = "hello"
x1
x1 + x2
```

A `list()`, unlike a matrix, can hold elements of different types or structures — here two matrices are stored together as list elements.

```r
matlist = list(x1 <- matrix(nrow = 2, ncol = 2, data = 1:4, byrow = TRUE), x2)
matlist
```

Single-bracket indexing `matlist[i]` returns a sub-list (still wrapped in `[[ ]]`), preserving the list structure.

```r
matlist[1]
matlist[2]
```

A list can mix entirely different object types together, such as a character vector, a numeric vector, and a matrix in one list.

```r
z1 = list( c("water", "juice", "lemonade"), rep(1:4, each = 2), matrix(data = 5:8, nrow = 2, ncol = 2, byrow = TRUE) )
z1
```

Double brackets `[[ ]]` extract the actual list element, while single brackets `[ ]` keep it wrapped as a list — so `z1[[1]][2]` reaches the value, but `z1[1][2]` returns `NULL`.

```r
z1[[1]]
z1[1][2]
z1[[1]][2]
```
